````python
def initialize_inventory(products) :
    inventory = {}
    for product in products:
        valid_input = False
        while not valid_input:
            try:
                quantity = int(input(f"Enter the quantity for {product}: "))
                if quantity < 0:
                    print("Quantity cannot be negative. Please enter a valid number.")
                else:
                    inventory[product] = quantity
                    valid_input = True
            except ValueError:
                print("Invalid input. Please enter a valid integer.")

    return inventory
````

````python
def get_customer_orders(inventory):
    valid_num_orders = False
    while not valid_num_orders:
        try:
            num_orders = int(input("Enter the number of customer orders: "))
            if num_orders < 0:
                print("Number of orders cannot be negative. Please enter a valid number.")
            else:
                valid_num_orders = True
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    customer_orders = set()
    for i in range(num_orders):
        valid_product = False
        while not valid_product:
            product = input(f"Enter the name of product #{i+1} that a customer wants to order: ")
            if product not in inventory:
                print(f"Error: '{product}' is not in the inventory. Please enter a valid product.")
            elif inventory[product] <= 0:
                print(f"Error: '{product}' is out of stock. Please choose another product.")
            else:
                customer_orders.add(product)
                valid_product = True
                
    return customer_orders
````

````python
def calculate_total_price(customer_orders):
    total_price = 0
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                price = float(input(f"Enter the price of {product}: "))
                if price < 0:
                    print("Price cannot be negative. Please enter a valid price.")
                else:
                    total_price += price
                    valid_price = True
            except ValueError:
                print("Invalid input. Please enter a numeric price.")
    return total_price
````

````python
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1
    inventory = {product: quantity for product, quantity in inventory.items() if quantity > 0}
    return inventory
````

````python
inventory = initialize_inventory(products)
````

````python
customer_orders = get_customer_orders(inventory)
````

````python
inventory = update_inventory(customer_orders, inventory)
````

````python
total_price = calculate_total_price(customer_orders)
print(f"\nTotal Price: {total_price}")
````
